```python
import zipfile
from pathlib import Path

# Path to zip on Google Drive
file_zip = '/content/drive/MyDrive/train.zip'
extract_dir = Path('/content/object_detection')

if extract_dir.exists() and any(extract_dir.iterdir()):
    print(f'Dataset already exists at: {extract_dir} -> skip unzip')
else:
    extract_dir.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(file_zip, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print(f'Unzipped {file_zip} to {extract_dir}')
```
